```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.normal(loc=170, scale=10, size=100)  # Mean = 170, Std Dev = 10, Sample Size = 100

#sample mean and standard error
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)  # Use ddof=1 for sample standard deviation
n = len(data)
standard_error = sample_std / np.sqrt(n)

# Confidence level (95%)
confidence_level = 0.95
z_score = stats.norm.ppf((1 + confidence_level) / 2)  # Two-tailed Z-score

#margin of error and confidence interval
margin_of_error = z_score * standard_error
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Confidence Interval (95%): {confidence_interval}")
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.hist(data, bins=15, alpha=0.6, color='g', edgecolor='black', density=True, label="Sample Data Distribution")

#confidence interval
plt.axvline(confidence_interval[0], color='red', linestyle='dashed', label=f'Lower Bound: {confidence_interval[0]:.2f}')
plt.axvline(confidence_interval[1], color='blue', linestyle='dashed', label=f'Upper Bound: {confidence_interval[1]:.2f}')
plt.axvline(sample_mean, color='black', linestyle='solid', label=f'Mean: {sample_mean:.2f}')

plt.title('Confidence Interval Visualization')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.show()
```
